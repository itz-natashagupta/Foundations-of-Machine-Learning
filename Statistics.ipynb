```python
# Z-Test
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n=len(sample)
```

```python
z_score = (sample_mean - population_mean)/(population_std/np.sqrt(n))
z_score
```

```python
p_value = 2 * (1 - norm.cdf(abs(z_score)))
```

```python
p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
#T-Test
import numpy as np
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
sample_mean = np.mean(sample)
sample_std = np.std(sample,ddof = 1)
n = len(sample)
mean_population = 170
```

```python
t_stats = (sample_mean - mean_population) / (sample_std / np.sqrt(n))
t_stats
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_stats),df = n -1 ))
alpha = 0.05
```

```python
if p_value < alpha:
    print("reject the null hypothesis")
else:
    print("accept the null hypothesis")
```

```python
# Chi - Square Test
import numpy as np
import seaborn as sns
from scipy.stats import chi2_contingency
import pandas as pd
```

```python
df = sns.load_dataset('titanic')
```

```python
df
```

```python
contingency_table = pd.crosstab(df['sex'],df['survived'])
```

```python
contingency_table
```

```python
chi2, p_value , dof , expected = chi2_contingency(contingency_table)
```

```python
expected
```

```python
chi2
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("we reject the null hypothesis and there is a significant relationship between gender and survivl")
else:
    print("there is no connenction")
```
